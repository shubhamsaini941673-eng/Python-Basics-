Question 2 : Explain the difference between List and Tuple in Python.

```python
1. Both List and Tuple are used to store multiple values in Python, but they have some important differences.

| Feature     | List                      | Tuple                              |
| ----------- | ------------------------- | ---------------------------------- |
| **Syntax**  | Uses `[ ]`                | Uses `( )`                         |
| **Mutable** | Yes, can be changed       | No, cannot be changed              |
| **Speed**   | Comparatively slower      | Comparatively faster               |
| **Use**     | Used when data may change | Used when data should remain fixed |
| **Example** | `my_list = [10, 20, 30]`  | `my_tuple = (10, 20, 30)`          |

Example

# List
numbers = [10, 20, 30]
numbers[0] = 50
print(numbers)   # [50, 20, 30]

# Tuple
numbers = (10, 20, 30)
# numbers[0] = 50  → Error

. In short: A List is mutable (changeable), while a Tuple is immutable (unchangeable).

2. Continue with Python basics

. Show common list methods
. Compare sets and dictionaries
```

Question 5 : Write a Python program to check whether a number is even or odd using conditional statements.

```python
1. Python Program to Check Whether a Number is Even or Odd

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is Even")
else:
    print("The number is Odd")

Example

Enter a number: 10
The number is Even

2. Explanation:

. input() takes a number from the user.
. int() converts the input into an integer.
. % (modulus) gives the remainder.
. If the remainder is 0, the number is Even; otherwise, it is Odd.
```

Question 8 : Write a Python program to remove duplicate values from a list using a set.

```python
1. Python Program to Remove Duplicate Values from a List Using a Set

numbers = [10, 20, 10, 30, 20, 40, 30]

unique_numbers = list(set(numbers))

print("Original list:", numbers)
print("List after removing duplicates:", unique_numbers)

Output

Original list: [10, 20, 10, 30, 20, 40, 30]
List after removing duplicates: [10, 20, 30, 40]

2. Explanation:

 1. A set stores only unique values, so duplicate values are automatically removed.

 2. set(numbers) converts the list into a set.

 3. list() converts the set back into a list.

Note: Sets do not guarantee the original order of elements.

3. Continue with Python basics

Preserve list order while removing duplicates
```

Question 9 : Write a function that returns the square of a number.

```python
1. Python Function to Return the Square of a Number

def square(num):
    return num * num

number = int(input("Enter a number: "))

result = square(number)

print("The square is:", result)

2. Output

Enter a number: 5
The square is: 25

3. Explanation:

. def square(num) creates a function named square.

. return num * num calculates and returns the square.

. The function is called using square(number).

4. Continue with Python functions

. Return cubes with a function
. Use a one-line square function
```

Question 10 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
1. Python Program to Count How Many Times a Number Appears in a List

numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print("The number", num, "appears", count, "times.")

2. Output

Enter the number to count: 2
The number 2 appears 3 times.

3. Explanation:

. numbers contains the given list.
. count() is used to count how many times a particular number appears.
. In this example, 2 appears 3 times.
```
